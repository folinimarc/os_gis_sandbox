## 3. Vectors, lists, and data frames
Vectors are core in R. Data frames are tabular data structures.

```python
# Vector
temps <- c(18.2, 19.5, 17.8, 20.1)
temps
mean(temps)

# List
profile <- list(user = "Marc", level = 1, active = TRUE)
profile
profile$user

# Data frame
df <- data.frame(
  city = c("Zurich", "Bern", "Basel"),
  population_m = c(0.43, 0.14, 0.18),
  stringsAsFactors = FALSE
)
df
str(df)
```

```python
# Optional: dplyr-style manipulation
library(dplyr)

df %>%
  mutate(population_k = population_m * 1000) %>%
  arrange(desc(population_k))
```

## 7. First geospatial example with sf
Create a tiny point dataset and transform CRS.

```python
library(sf)

pts <- data.frame(
  name = c("Zurich", "Bern", "Basel"),
  lon = c(8.5417, 7.4474, 7.5886),
  lat = c(47.3769, 46.9480, 47.5596)
)

pts_sf <- st_as_sf(pts, coords = c("lon", "lat"), crs = 4326)
pts_sf

# Reproject to Swiss projected CRS (CH1903+ / LV95)
pts_lv95 <- st_transform(pts_sf, 2056)
pts_lv95
```

```python
plot(st_geometry(pts_sf), pch = 19, col = "firebrick", main = "Sample Points (WGS84)")
```
